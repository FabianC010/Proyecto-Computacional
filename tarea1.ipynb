##Problema 1: Funciones Caracteristicas
Las funciones caracterısticas en teorıa de probabilidad contienen la misma informacion que las PMFs/PDFs, sin embargo, son
increıblemente utiles.  


*   Investigue que es una funcion caracterıstica en teorıa de probabilidad, sus propiedades y sus usos.
*   Calcule la funcion caracterıstica de la variable aleatoria X ∼ Bern(p) de Bernoulli que vimos en clase
*   Usando esta funcion caracterıstica, calcule todos los momentos E[X^n] de la variable aleatoria de Bernoulli, para todo
n >1.

*    Realice lo mismo para la distribucion normal.



Una funcion caracteristica es una funcion la cual tiene como variable una variable aleatoria real pero la funcion devuelve numero complejos.
$$
\phi_{X}t=E(e^{itX}) = \int_{-∞}^{+\infty}e^{itx}f_{X}(x)dx
$$
donde $f_X$ es la funcion de densidad de probabilidad

```python
import numpy as np

Bern = np.random.binomial(1.0, 0.5, 10000) #creamos la variable aleatoria de bernoulli como la variable aleatoria binomial con n=1
t = np.linspace(0,10, np.size(Bern)) #creamos una t que inicia y termina en un intervalo arbitrario con el mismo tamaño que la variable aleatoria de Bernoulli
FuncCaracteristic = np.mean(np.exp(1j*t*Bern)) #Calculamos su funcion caracteristica utilizando mean de numpy
print(FuncCaracteristic)
```

pero la funciond de densidad de probabilidad de la distribucion binomial es
$$
P(X = i) =\binom{n}{i} \, p^{i} (1 - p)^{n - i}
$$
por lo que la funcion caracteristica es
$$
\phi_{X}t =  ∑ e^{itx} \binom{n}{i} \, p^{i} (1 - p)^{n - i}
$$
cuya solucion es
$$
\phi_{X}t = (1-p)+pe^{it}
$$
cuyos momentos corresponden a las derivaadas de la funcion caracteristicas evaluadas en el origen.


```python
p=0.5
t=0
FuncCaracteristic1 = (1-p)+p*np.exp(1j*t)
print(FuncCaracteristic1)
FuncCaracteristic2 = (1j*p*np.exp(1j*t))
print(FuncCaracteristic2)
FuncCaracteristic3 = (-p*(np.exp(1j*t)))
print(FuncCaracteristic3)
for i in range(1,100):
  FuncCaracteristicn = (1j**i)*p*np.exp(1j*t)
  print(FuncCaracteristicn)
#Podemos ver que el valor para todo n>1 es igual a 0.5 o "p"
```

1.Normalizamos la funcion despejando a, y tomando $x_1 = x$ y $x_2 = y$, solamente por conviencia.

 Prob(f(x,y)) = \frac{1}{\pi a^4} \int _{-\infty}^{+\infty} \int _{-\infty}^{+\infty}  ( {y - x} )^2 \exp\left[ -\frac{x^2 + y^2}{a^2} \right] dxdy =1

2. Solo podemos determinar si son estadisticamente independientes si el producto de las distribuciones marginales es igual a la funcion original osea:

3. La probabilidad condicional esta dada por:
$$
P(A|B) = \frac{P(AB)}{P(B)}= \frac{P(f(x,y))}{P(f_Y(y))}  
$$
Pero la funcion de probabilidad condicional esta dada por
$$
\frac{f(x,y)}{f_Y(y)}
$$
Primero
$$
P(A|B) = \frac{2 a^2 \sqrt{π}}{e^{-y^2/a^2} (a^2 - 2 y^2) |a|}
$$
y segundo
$$
\frac{f(x,y)}{f_Y(y)} = \frac{ \frac{1}{\pi a^2} \left( \frac{x_2 - x_1}{a} \right)^2 \exp\left[ -\frac{x_1^2 + x_2^2}{a^2} \right]}{\frac{2 a^2 \sqrt{π}}{e^{-y^2/a^2} (a^2 - 2 y^2) |a|}}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

def JointPDF(x, y, a):
    Coeficiente = 1 / (np.pi * a**2)
    TerminoCruzado= ((y - x) / a)**2
    Exponente = -(x**2 + y**2) / a**2

    # Joint PDF
    return Coeficiente * TerminoCruzado * np.exp(Exponente)
def MarginalPDF(y, a):
    Coeficiente = 1 / (2 * a**2 * np.sqrt(np.pi))
    Exponente = -(y**2) / a**2
    TerminoY = (a**2 - 2 * y**2) * abs(a)

    # Marginal PDF
    return Coeficiente * np.exp(Exponente) * TerminoY

#La funcion de probabilidad condicional es
def ConditionalPDF(x, y, a):
    return JointPDF(x, y, a) / MarginalPDF(y, a)

y = [0, 1/2, 1, 2]
x = np.linspace(-10, 10, 10000)
a = 1.0
for i in y:
  PDFCondicional = ConditionalPDF(x, i, a)
  plt.plot(x, PDFCondicional)
  plt.title("y = "+str(i))
  plt.show()

```

##Problema 4 TEORIA CINETICA DE GASES
Considere una partıcula en un gas ideal 3D, con masa m = 1 y kBT = 1. Cada componente de velocidad de la partıcula es
muestreada de una distribucion normal

$$
v_x, v_y, v_z \sim \mathcal{N}(0, \sigma^2), \tag{3}
$$
donde consideramos
$\sigma^2 = k_B T / m = 1$ y

$$
\mathcal{N}(\mu, \sigma^2) = \frac{1}{\sigma \sqrt{2\pi}} e^{-(x - \mu)^2 / 2\sigma^2} \tag{4}
$$
es la distribución normal con valor de expectación
$\mu$ y varianza
$\sigma^2$. La energía cinética total de la partícula es

$$
T = \frac{1}{2}(v_x^2 + v_y^2 + v_z^2). \tag{5}
$$
Vamos a verificar si se cumple la ley débil y fuerte de los números grandes en este problema.


1.Demuestre que
$E[T] = 3/2$


2.Realice un histograma de la energía cinética usando
$N = 5, 50, 1000$ muestras. Analice el resultado con base en la ley débil de los grandes números. Para esto puede ser muy útil numpy.histogram y matplotlib.pyplot.hist


3.Realice un gráfico del promedio acumulativo de la energía cinética con respecto al número de muestras. Es decir, realice un promedio de la energía cinética con respecto a
$N$. Para esto puede ser muy útil numpy.cumsum. Analice el resultado con respecto a la ley fuerte de los grandes números.




```python
import numpy as np
import matplotlib.pyplot as plt
N=10000000
vx, vy, vz = np.random.normal(0, 1.0, N), np.random.normal(0, 1.0, N), np.random.normal(0, 1.0, N) #para cada componente asignamos una distribucion de velocidades normales diferente para cada una.
MeanT = 1/2*np.sum(vx**2+vy**2+vz**2)/N #De un solo computamos el promedio (mean) para ahorrar tiempo de calculo
print(MeanT)

N = [5, 50, 1000, 100000]
for i in N:
  T = np.zeros(i)
  vx, vy, vz = np.random.normal(0, 1.0, i), np.random.normal(0, 1.0, i), np.random.normal(0, 1.0, i)
  T = 1/2*(vx**2+vy**2+vz**2)
  plt.hist(T, density = True, bins = 100)
  plt.title("N = "+str(i))
  plt.show()
```

```python
N = [5,50,1000,10000]
for i in N:
  T = np.zeros(i)
  vx, vy, vz = np.random.normal(0, 1.0, i), np.random.normal(0, 1.0, i), np.random.normal(0, 1.0, i)
  T = np.cumsum( 1/2*(vx**2+vy**2+vz**2))/i
  print(T[-1])
  plt.plot(T)
  plt.title("N = "+str(i))
  plt.show()
```
